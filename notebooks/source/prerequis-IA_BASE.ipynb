## 3 Fonctions et structure conditionnelle

<font color='red'>**Q4:**</font> Définissez une fonction ```sign(x)``` qui renvoie ```0``` si ```x``` est nul, ```1``` si ```x``` est positif et ```-1``` sinon. On pourra s'aider de structure de contrôle ```if then else```. Évaluez ```sign(420)```, ```sign(-42)``` et ```sign(0)```

````python
def sign(x):
    if x==0:
        return 0
    else:
        if x>0:
            return 1
        else:
            return -1

sign(420), sign(-42), sign(0)
````

<font color='red'>**Q5:**</font> Définissez une fonction```volume(h,r)```  renvoyant le volume d'un cylindre, en décomposant ce calcul avec des variables locales comme suit:
$$\mathrm{Surface}_{\mathrm{disque}}=3.14*r*r$$
$$\mathrm{Volume}_{\mathrm{cylindre}}= h * \mathrm{Surface}_{\mathrm{disque}}$$

````python
def volume(h,r):
    s_disque = 3.14 * r * r
    return h * s_disque
````
