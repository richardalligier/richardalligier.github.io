## 1 Une (très) courte introduction à Jupyter Notebook
Les Jupyter notebook sont des fichiers `.ipynb` contenant des cellules. Il y a deux types de cellule: 
1. Cellules Markdown contenant du texte;
2. Cellules Code contenant du code Python executable;

Les cellules de code peuvent être édités. Les différentes cellules de code partagent le même "namespace". Pour exécuter une cellule de code, vous avez deux raccourcis clavier:

- `Ctrl`+`Return`: exécute la cellule courante;
- `Shift`+`Return`: exécute la cellule courante, et la cellule suivante d'en dessous devient cellule courante.
